```python
### setup core variables
n_conditions = 8   # n rows in conditions file
tot_trials = 720    # must be a multiple of n_conditions
interrupt_freq = 10 / 100    # percentage of interruption trials
n_blocks = 6

trials_per_block = int(tot_trials / n_blocks)   # this number must be a factor of tot_trials AND must be divisible by n_conditions
if trials_per_block % n_conditions != 0:
    raise Exception(f'n of trials per block must be divisible by {n_conditions}')

tot_interruptions = round(tot_trials * interrupt_freq)  #- 2  # ensure tot n interruptions is NOT a factor of tot_trials, otherwise last trial in each block will be an interruption trial
                                                        # but is this an issue for me? given my psychopy structure, i will always have a post interruption trial even if it's the last row of the block

interrupt_per_block = int(tot_interruptions / n_blocks) # must be a multiple of 2 or 3
if interrupt_per_block % 2 != 0 and interrupt_per_block % 3 != 0:
    raise Exception("n of interruptions per block must be a  multiple of 2 or 3")

if tot_interruptions == (interrupt_per_block * n_blocks):
    tot_interruptions = tot_interruptions
else:
    print(f'changed tot_interuptions! was originally {tot_interruptions}')
    tot_interruptions = interrupt_per_block * n_blocks

avg_gap = int(trials_per_block / interrupt_per_block)   # average n trials between evenly-spaced interruptions

gap_range = list(range((avg_gap - 2), (avg_gap + 2) + 1))   # add ±2 jitter to gaps between interruptions - NB: the n of items in gap_range must be a factor of interrupt_per_block
if interrupt_per_block % len(gap_range) == 0:
    gap_range = gap_range
elif interrupt_per_block % 3 == 0:
    gap_range = [(avg_gap - 2), avg_gap, (avg_gap + 2)]
else:
    gap_range = list(range((avg_gap - 2), (avg_gap + 1) + 1))

print(
    f' n conditions =  {n_conditions} \n tot trials = {tot_trials} \n interrupt frequency = {interrupt_freq} \n n blocks = {n_blocks} \n trials per block = {trials_per_block}'
    f'\n tot interruptions = {tot_interruptions} \n interruptions per block = {interrupt_per_block} \n average gap = {avg_gap} \n gap range = {gap_range}')
```

```python
# shuffle trials in chunks of 16 and append to new dataframe to make entire list of evenly randomised trials

trials_list = [] # initialise empty list to store shuffled trials

num_repeats = int(trials_per_block / n_conditions) # final total trial n IN EACH BLOCK

for block in range(n_blocks):
    for i in range(num_repeats):
        selected_conditions = []

        for r in range(n_conditions):
            pick_rows = random.choice([r, r + n_conditions])
            selected_conditions.append(conditions.iloc[pick_rows])
        
        selected_df = pd.DataFrame(selected_conditions)
        shuffled_conditions = selected_df.sample(frac=1).reset_index(drop=True)  # randomly shuffle the selected 16 conditions
        trials_list.append(shuffled_conditions)  # append the shuffled chunk to trial list

trials_df = pd.concat(trials_list, ignore_index=True) # transform trial list into trial df
#trials_df.head(16*num_repeats)  # show entire df
```

```python
# Generate and evenly shuffle possible interruption lengths
lengths = []

if interrupt_per_block % 2 == 0:
    for block in range(n_blocks):   # create new shuffled list per block
        block_lengths = [1] * int(interrupt_per_block/2) + [2] * int(interrupt_per_block/2) # interruption task can be 1 or 2 sums long
        random.shuffle(block_lengths)
        lengths.append(block_lengths)   # compile all blocks into one big list of lengths
    print(lengths)
    print(len(lengths)) # should = n_blocks

elif interrupt_per_block % 3 == 0:
    for block in range(n_blocks):   # create new shuffled list per block
        block_lengths = [1] * int(interrupt_per_block/3) + [2] * int(interrupt_per_block/3) + [3] * int(interrupt_per_block/3) # interruption task can be 1, 2, or 3 sums long
        random.shuffle(block_lengths)
        lengths.append(block_lengths)   # compile all blocks into one big list of lengths
    print(lengths)
    print(len(lengths)) # should = n_blocks
```

```python
# Repeat the gap_range enough to evenly cover all interruptions
gaps = []
for block in range(n_blocks):
    block_gaps = gap_range * (interrupt_per_block // len(gap_range)) # create a list as long as n_interruptions that contains equal repeats of each number in gap_range
    print(block_gaps)
    random.shuffle(block_gaps)
    gaps.append(block_gaps)
print(gaps)
print(len(gaps))
```

```python
# Initialise empty lists to store interruption trials and corresponding lengths
interruption_trials = []
interruption_lengths = []
add_block_counter = [(trials_per_block * 0), (trials_per_block * 1), (trials_per_block * 2), (trials_per_block * 3), (trials_per_block * 4), (trials_per_block * 5)]

for block in range(n_blocks):   # pick interruption trial idx for each block
    current_trial = gaps[block][-1] + (trials_per_block * block)    # at the start of each block, pick a new trial idx to assign first interruption.
                                    # in gaps, in the list corresponding to the current block nr, pick the last item.
                                    # cannot pick the first item or a random or fixed number because it will mess up the balance of gaps in that list/block, 
                                    # and might cause eventually fewer tot interruptions than planned.
                                    # need to pick last item because it's the only one that isn't used for updating the current_trial counter.
    block_interruption_trials = []
    block_interruption_lengths = []

    for idx_i in range(interrupt_per_block):    # iterate over the nr of trials per block
        if current_trial > int(trials_per_block + (trials_per_block * block)):
            break   # stop loop at the end of each block
        block_interruption_trials.append(current_trial)             # store current interruption trial idx
        block_interruption_lengths.append(lengths[block][idx_i])    # store current interruption length
        current_trial += gaps[block][idx_i]                         # update counter by adding current gap to current trial idx
    interruption_trials.extend(block_interruption_trials)
    interruption_lengths.extend(block_interruption_lengths)

print(interruption_trials)
print(len(interruption_trials))
print(interruption_lengths)
print(len(interruption_lengths))
```

```python
# Create a tuple to store all the lists made above
interruption_data = list(zip(interruption_trials, interruption_lengths, trial_saliency))
interruption_data.sort()  # Sort by trial index
print(interruption_data)
print(len(interruption_data))
```

```python
# Initialise tuple (?) to store trials
trial_data = []

# Initialise tracking how many trials since last interruption
since_last_interrupt = None

# Create a list of trial numbers where interruptions will happen
next_interrupts = [x[0] for x in interruption_data]

# Start at the first interruption in the list
interrupt_idx = 0
```

```python
# Loop through every trial
for trial in range(1, tot_trials + 1):
    
    # Set default values
    is_interruption = 0
    interrupt_length = 0
    interrupt_saliency = None
    take_break = False
    
    # check if interruption trial matches with current trial 
    if interrupt_idx < len(interruption_data) and interruption_data[interrupt_idx][0] == trial: 
        is_interruption = 1
        interrupt_length = interruption_data[interrupt_idx][1]          # get length
        interrupt_saliency = interruption_data[interrupt_idx][2]        # get trial saliency
        since_last_interrupt = 1                                        # reset timer     
        interrupt_idx += 1                                              # move to next interruption in the list
    else:
        if since_last_interrupt is not None:        # if an interruption has occured before
            since_last_interrupt += 1               # increase counter since last interruption

    # create a list of upcoming interruptions that happen after current trial
    future_interrupts = [i for i in next_interrupts if i > trial]

    # If there are future interruptions calculate how many trials away the next interruption is
    if future_interrupts:
        until_next_interrupt = trial - future_interrupts[0]
    else:
        until_next_interrupt = None
    
    # check if it's time for a block break every 144 trials
    if trial % trials_per_block == 0:
        take_break = True

    # convert data into a table structure
    trial_data.append({
        'trial_nr': trial,
        'is_interruption_trial': is_interruption,
        'interruption_duration': interrupt_length,
        'post_interrupt_saliency': interrupt_saliency,
        'since_last_interr': since_last_interrupt,
        'until_next_interr': until_next_interrupt,
        'time_to_break' : take_break
    })

# Save to csv
df = pd.DataFrame(trial_data)
df.to_csv("my_interruptions_list.csv", index=False)
```
